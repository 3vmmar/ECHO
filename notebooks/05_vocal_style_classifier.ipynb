# Vocal-style classifier

Phase 5. Classifies the expressed vocal style of acted clips as happy, sad, angry or neutral. The label describes the voice, never the speaker's feelings.

- **Inputs:** emotion clips and labels
- **Outputs:** `results/tables`, `results/figures`
- **Rule:** scaling and SVM settings are fit on training speakers and chosen on validation speakers; test speakers are scored once.

## 5. Test results

Four-class macro-F1, per-class precision and recall and a confusion matrix, against a class-frequency dummy baseline. Exploratory target: macro-F1 of at least 0.55.

## 6. Limits

Acted expressions and shared sentence texts limit conclusions to unseen speakers performing this controlled task; accuracy on natural diary speech is not validated.
