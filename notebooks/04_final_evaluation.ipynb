# Final evaluation

Phase 4. One run of A, B and C on the six held-out test speakers. Models, scoring and decoding do not change after this point.

- **Inputs:** test predictions and `configs/frozen.json`
- **Outputs:** `results/tables`, `results/figures`
- **Rule:** every outcome is reported, including a missed target.

## 1. Pre-run checklist

`configs/frozen.json` exists, and the scorer version, protocol hash and checkpoint hash match it.

## 2. Pooled results

Strict WER, relaxed WER and CER for A, B and C.

## 3. Per-speaker results

Paired table for the six test speakers.

## 4. Paired bootstrap

2,000 resamples of whole test speakers; C minus A and C minus B with 95% percentile intervals. Exploratory, not a significance claim.

## 5. Success criteria

At least a 10% relative reduction in pooled strict WER for C versus A, improvement on at least 5 of the 6 test speakers, and a positive reduction under relaxed scoring.

## 6. Error analysis

Dialect and code-switching error categories; mixed-language utterances reported separately.
