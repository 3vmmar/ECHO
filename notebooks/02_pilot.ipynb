# Pilot

Phase 2. One or two pilot speakers, never test speakers. Confirms that system B runs under the fixed protocol, measures transcription time and drafts the scoring rules.

- **Inputs:** pilot clips and their references
- **Outputs:** draft lists in `configs/scoring`, entries in `docs/decisions.md`
- **Rule:** pilot data only; references are typed without ASR drafts.
