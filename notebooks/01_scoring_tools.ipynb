# Scoring tools

Phase 1. Shows the single scoring function on synthetic and development examples. The scorer is frozen before any test inference.

- **Inputs:** synthetic strings, pilot and validation references
- **Outputs:** none; decisions go to `docs/decisions.md`
- **Rule:** never use test references or predictions here.
